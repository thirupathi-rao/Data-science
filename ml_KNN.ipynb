```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Original data
data = {
    "No. of sales": [100, 120, 80, 140, 190, 200, 5000],
    "Sale Value": [1000, 1100, 900, 1500, 2000, 2100, 100000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Initialize the scaler
scaler = StandardScaler()

# Fit and transform the data
standardized_data = scaler.fit_transform(df)

# Convert back to DataFrame for readability
standardized_df = pd.DataFrame(standardized_data, columns=df.columns)

print("Original Data:")
print(df)
print("\nStandardized Data:")
print(standardized_df)
```

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler

# Original data
data = {
    "No. of sales": [100, 120, 80, 140, 190, 200],
    "Sale Value": [1000, 1100, 900, 1500, 2000, 2100]
}

# Create DataFrame
df = pd.DataFrame(data)

# Initialize RobustScaler
scaler = RobustScaler()

# Fit and transform the data
robust_scaled_data = scaler.fit_transform(df)

# Convert to DataFrame
robust_scaled_df = pd.DataFrame(robust_scaled_data, columns=df.columns)

print("Original Data:")
print(df)
print("\nRobust Scaled Data:")
print(robust_scaled_df)
```
