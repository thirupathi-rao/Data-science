```python
import cv2
import numpy as np
import mediapipe as mp
```

```python
fd=mp.solutions.face_detection

face_det=fd.FaceDetection(min_detection_confidence=0.5, model_selection=0)

img=cv2.imread(r"D:\Downloads\me.jpg")
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
rgb=img[:,:,::-1]#easy way to convert to rgb by reversing bgr umg

output=face_det.process(rgb)

output.detections

ih,iw,d=img.shape

x=int(output.detections[0].location_data.relative_bounding_box.xmin*iw)
y=int(output.detections[0].location_data.relative_bounding_box.ymin*ih)
w=int(output.detections[0].location_data.relative_bounding_box.width*iw)
h=int(output.detections[0].location_data.relative_bounding_box.height*ih)

img1=img.copy()
cv2.rectangle(img1,(x,y),(x+w,y+h),(0,255,0),2)

cv2.imshow("f",img1)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
img1=img.copy()
for i in range(0,6):
    x=int(output.detections[0].location_data.relative_keypoints[i].x*iw)
    y=int(output.detections[0].location_data.relative_keypoints[i].y*ih)
    cv2.circle(img1,(x,y),4,(0,255,0),-1)
    cv2.putText(img1,fd.FaceKeyPoint(i).name,(x-10,y-10),cv2.FONT_HERSHEY_COMPLEX,0.5,(255,0,0),1)
cv2.imshow("f",img1)
cv2.waitKey()
cv2.destroyAllWindows()
```

# using mediapipe for drawing 

```python
img1=img.copy()
mp.solutions.drawing_utils.draw_detection(image=img1,detection=output.detections[0],
                                         keypoint_drawing_spec=mp.solutions.drawing_utils.DrawingSpec(
                                             color= (255, 0, 0),thickness= 2,circle_radius= 2,),
                                        bbox_drawing_spec=mp.solutions.drawing_utils.DrawingSpec(
                                             color= (0, 255, 0),thickness= 2,circle_radius = 2,))
cv2.imshow("f",img1)
cv2.waitKey()
cv2.destroyAllWindows()
```

# with video

```python
fd=mp.solutions.face_detection
vid=cv2.VideoCapture(0)
with fd.FaceDetection(min_detection_confidence=0.5, model_selection=1) as face_det:
    while True:
        s,f=vid.read()
        if s==False:
            break
        rgb=f[:,:,::-1]
        output=face_det.process(rgb)
        
        if output.detections:
            mp.solutions.drawing_utils.draw_detection(image=f,detection=output.detections[0])
        cv2.imshow("f",f)
        if cv2.waitKey(1)==ord('q'):
            break

cv2.destroyAllWindows()
vid.release()
        
```

# face mesh

```python
fp=mp.solutions.face_mesh
fm_model=fp.FaceMesh(static_image_mode=True,
    max_num_faces=1,
    refine_landmarks=False,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5,)
img=cv2.imread(r"D:\Downloads\me.jpg")
img=cv2.resize(img,(1080,1080))
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
rgb=img[:,:,::-1]#easy way to convert to rgb by reversing bgr umg

result=fm_model.process(rgb)
if result.multi_face_landmarks:
    for i in result.multi_face_landmarks:
        mp.solutions.drawing_utils.draw_landmarks(image=img,landmark_list=i)
cv2.imshow("a",img)
cv2.waitKey()
cv2.destroyAllWindows()
```

# face mesh video

```python
fp=mp.solutions.face_mesh
fm_model=fp.FaceMesh(static_image_mode=True,
    max_num_faces=2,
    refine_landmarks=False,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5,)
vid=cv2.VideoCapture(0)
while True:
    b,f=vid.read()
    if b == False:
        break
    rgb=f[:,:,::-1]
    result=fm_model.process(rgb)
    if result.multi_face_landmarks:
        for i in result.multi_face_landmarks:
            mp.solutions.drawing_utils.draw_landmarks(image=f,landmark_list=i,connections=fp.FACEMESH_TESSELATION,
                                                      landmark_drawing_spec=mp.solutions.drawing_utils.DrawingSpec(
                                                          color= (0, 255, 0),thickness= 1,circle_radius = 1,),
                                                      connection_drawing_spec=mp.solutions.drawing_styles.DrawingSpec(
                                                          color= (255,0, 0),thickness= 1,circle_radius = 1,))
    cv2.imshow("a",f)
    if cv2.waitKey(1)==ord("q"):
        break
cv2.destroyAllWindows()
vid.release()
```
