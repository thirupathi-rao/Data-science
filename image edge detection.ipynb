```python
import cv2

# Load image once
original = cv2.imread(r"D:\Downloads\mixkit-people-in-the-subway-hall-in-tokyo-4454-hd-ready - frame at 0m2s.jpg")

cv2.namedWindow("f1")
cv2.createTrackbar("low", "f1", 50, 255, lambda x: None)
cv2.createTrackbar("high", "f1", 150, 255, lambda x: None)

while True:
    img = original.copy()
    gs = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    bi = cv2.GaussianBlur(gs, (5, 5), 1)

    l = cv2.getTrackbarPos("low", "f1")
    h = cv2.getTrackbarPos("high", "f1")

    canny = cv2.Canny(bi, l, h)

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
    canny = cv2.morphologyEx(canny, cv2.MORPH_CLOSE, kernel, iterations=1)
    canny = cv2.dilate(canny, kernel, iterations=1)

    c, _ = cv2.findContours(canny, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    sc = sorted(c, key=cv2.contourArea, reverse=True)

    combined = None
    for i in sc:
        if cv2.contourArea(i) < 500:
            continue
        if combined is None:
            combined = cv2.boundingRect(i)
        else:
            x, y, w, h = cv2.boundingRect(i)
            x0 = min(combined[0], x)
            y0 = min(combined[1], y)
            x1 = max(combined[0] + combined[2], x + w)
            y1 = max(combined[1] + combined[3], y + h)
            combined = (x0, y0, x1 - x0, y1 - y0)
    
    if combined:
        x, y, w, h = combined
        cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)


    cv2.imshow("f1", img)
    if cv2.waitKey(1) == ord("q"):
        break

cv2.destroyAllWindows()
```

# gpt pro code

```python
import cv2

vid=cv2.VideoCapture(r"D:\Downloads\mixkit-people-in-the-subway-hall-in-tokyo-4454-hd-ready.mp4")
hog = cv2.HOGDescriptor()
hog.setSVMDetector(cv2.HOGDescriptor_getDefaultPeopleDetector())

while True:
    ret, frame = vid.read()
    if not ret:
        break

    frame = cv2.resize(frame, (640, 360))  # Resize for faster processing
    boxes, weights = hog.detectMultiScale(frame, winStride=(8, 8))

    for (x, y, w, h) in boxes:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv2.imshow("People Detection", frame)

    if cv2.waitKey(30) == ord("q"):
        break

vid.release()
cv2.destroyAllWindows()
```

# approxPolydp

```python
import cv2
import numpy as np

img = cv2.imread(r"D:\shapes_and_colors.jpg")
img1 = img.copy()

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 1)
canny = cv2.Canny(blur, 100, 170)

contours, _ = cv2.findContours(canny, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

for c in contours:
    approx = cv2.approxPolyDP(c, 0.03 * cv2.arcLength(c, True), True)
    cv2.drawContours(img, [approx], -1, (0, 255, 0), 1)

    # Centroid for text position
    M = cv2.moments(c)
    if M["m00"] != 0:
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])
    else:
        cx, cy = approx.ravel()[0], approx.ravel()[1]

    # Shape classification
    shape = "unknown"
    sides = len(approx)
    if sides == 3:
        shape = "triangle"
    elif sides == 4:
        x, y, w, h = cv2.boundingRect(approx)
        aspect_ratio = w / float(h)
        shape = "square" if 0.95 < aspect_ratio < 1.05 else "rectangle"
    elif sides == 5:
        shape = "pentagon"
    elif sides == 6:
        shape = "hexagon"
    elif sides > 6:
        shape = "circle"

    # Optional: Color detection at center
    b, g, r = img[cy, cx]
    color = "unknown"
    if r > 150 and g < 100 and b < 100:
        color = "red"
    elif g > 150 and r < 100 and b < 100:
        color = "green"
    elif b > 150 and r < 100 and g < 100:
        color = "blue"

    label = f"{color} {shape}"
    cv2.putText(img, label, (cx, cy), cv2.FONT_HERSHEY_COMPLEX, 0.5, (255, 0, 255), 1)

cv2.imshow("Original", img1)
cv2.imshow("Detected", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
