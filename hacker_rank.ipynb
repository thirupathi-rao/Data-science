```python
from collections import Counter

s = input().strip()

# Count character frequencies
count = Counter(s)

# Sort: First by -count (descending), then by character (ascending)
sorted_items = sorted(count.items(), key=lambda x: (-x[1], x[0]))

# Print top 3
for char, freq in sorted_items[:3]:
    print(f"{char} {freq}")
```

```python
s = input().strip()
us=set(s)
l=[]
for k in us:
    count=sum(1 for j in s if k==j)
    l.append([k,count])
l=sorted(l,key=lambda x: (-x[1], x[0]))
for i in l[:3]:
    print(f"{i[0]} {i[1]}")
```
