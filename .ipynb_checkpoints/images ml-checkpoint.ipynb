```python
import cv2
import numpy as np

# Create a blank image
canvas = np.full((500, 500, 3),255, dtype=np.uint8)
drawing = False  # True if the mouse is pressed
last_point = None  # To draw continuous lines

# Mouse callback function
def draw_on_hover(event, x, y, flags, param):
    global drawing, last_point

    if event == cv2.EVENT_MOUSEMOVE:
        if drawing:
            if last_point is not None:
                cv2.line(canvas, last_point, (x, y), (0, 0, 0), 6)
            last_point = (x, y)
        else:
            # Show position on hover
            temp = canvas.copy()
            cv2.circle(temp, (x, y), 5, (0, 255, 0), -1)
            #cv2.imshow("w1", temp)

    elif event == cv2.EVENT_LBUTTONDOWN:
        drawing = True
        last_point = (x, y)

    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False
        last_point = None
    elif event ==  cv2.EVENT_RBUTTONDOWN :
        cv2.circle(canvas,(x,y),50,(0,0,255),5)
        last_point = (x, y)

# Create a named window
cv2.namedWindow("w1")
cv2.setMouseCallback("w1", draw_on_hover)

frames=[]
c=0
while True:
    cv2.imshow("w1",canvas)
    if cv2.waitKey(5) == ord("s"):
        frames.append(canvas)
        print("saved")
        c+=1
        canvas=np.full((500, 500, 3),255, dtype=np.uint8)
    if cv2.waitKey(5) == ord("c"):
        break
cv2.destroyAllWindows()
```

# bitwise operations

```python
can1=np.zeros((500,500),dtype=np.uint8)
cv2.circle(can1,(250,250),200,(255),-1)
```

```python
can2=np.zeros((500,500),dtype=np.uint8)
cv2.rectangle(can2,(100,100),(400,400),(255),-1)
```

```python
b_and=cv2.bitwise_and(can1,can2,)
```

```python
cv2.imshow("1",can1)
cv2.imshow("2",can2)
cv2.imshow("3",b_and)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
b_or=cv2.bitwise_or(can1,can2,)
```

```python
cv2.imshow("1",can1)
cv2.imshow("2",can2)
cv2.imshow("3",b_or)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
b_xor=cv2.bitwise_xor(can1,can2,)
```

```python
cv2.imshow("1",can1)
cv2.imshow("2",can2)
cv2.imshow("3",b_xor)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
b_not=cv2.bitwise_not(can1,can2,)
```

```python
cv2.imshow("1",can1)
cv2.imshow("2",can2)
cv2.imshow("3",b_not)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# and = intersection, xor= non intersectiong ,or= both,not= colour reversal (black to white and vise versa)
```

## bitwise and with mask

```python
img = cv2.imread(r"D:\Downloads\low res.jpg")
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
lower_blue = np.array([10, 50, 50])
upper_blue = np.array([140, 255, 255])
mask = cv2.inRange(hsv, lower_blue, upper_blue)

result = cv2.bitwise_and(img, img, mask=mask)


cv2.imshow("Original", img)
cv2.imshow("Blue Mask", mask)
cv2.imshow("Filtered Blue Areas", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
np.unique(mask)
```

```python
def show_hsv_values(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        pixel = hsv[y, x]
        print(f"HSV at ({x}, {y}): {pixel}")

img = cv2.imread(r"D:\Downloads\low res.jpg")
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow("Click to check HSV", img)
cv2.setMouseCallback("Click to check HSV", show_hsv_values)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
