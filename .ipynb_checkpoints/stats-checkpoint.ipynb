```python
from itertools import product
a = list(product([1, 2, 3, 4, 5, 6], repeat=2))
b=[[] for _ in range(6)]
for i in range(6):
    for j in range(6):
        b[i].append(a[(i*6)+j])
        
print(b)
```

```python
from itertools import product

# Generate all possible pairs of dice rolls
a = list(product([1, 2, 3, 4, 5, 6], repeat=2))

# Create an empty list b with 6 sublists
b = [[] for _ in range(6)]

# Fill the sublists in b with the appropriate pairs from a
for i in range(6):
    for j in range(6):
        b[i].append(a[(i * 6) + j])

# Print the result
print(b)
```

```python
df.groupby(['Temperature','Play'])["Play"].count()
```

```python
pd.crosstab(df['Temperature'],df['Play'])
```

```python
(pd.crosstab(df['Temperature'],df['Play'])/len(df))*100
```

```python
(pd.crosstab(df['Temperature'],df['Play'],margins=True)/len(df))*100
```

```python
pd.crosstab(df['Temperature'],df['Play'],normalize = "index",margins=True)
```

```python
pd.crosstab(df['Outlook'],df['Play'],margins=True)/len(df)
```

```python
pd.crosstab(df['Outlook'],df['Play'],normalize="index")
```

```python
pd.crosstab(df['Outlook'],df['Play'],normalize="columns")
```
