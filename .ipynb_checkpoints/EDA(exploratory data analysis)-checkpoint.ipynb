```python
l1 = [1, 2, 3]
l2 = [4, 5, 6]
ans = []

for x, y in zip(l1, l2):
    ans.append(x + y)

print(ans)
```

The zip function pairs elements from l1 and l2 together:
(1, 4), (2, 5), (3, 6)
For each pair (x, y), the code adds x and y together and appends the result to ans.

```python
a=np.arange(5,25).reshape(5,4)
(x,y)=a.shape
l=[]

for i in range(0,x):
    for j in range(0,y):
        if a[i,j]%2==0:
            if a[i,j]%5==0:
                l.append('hurry')
            else:
                l.append('oops')
        else:
            if a[i,j]%5==0:
                l.append('hi')
            else:
                l.append('bye')
        
b=np.array(l).reshape(x,y)
print(b)
```

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
for x in np.nditer(arr):
    print(x)
```

```python
arr = np.array([[1, 2], [3, 4]])
for x in np.nditer(arr, op_flags=['readwrite']):
    x[...] = x * 2  # Modifying the array in place
print(arr)
```

```python
a = np.array([1, 2, 3])
b = np.array([[10], [20], [30]])
for x, y in np.nditer([a, b]):
    print(x, y)
```

```python
arr = np.array([[1, 2], [3, 4]])
for x in np.nditer(arr, order='F'):
    print(x)
```

# Summary of Key Arguments

op_flags: Controls whether the array is read-only (['readonly']), writeable (['readwrite']), or write-only (['writeonly']).

flags: Controls iteration behavior. Common flags include:
- `'c_index'`: Track the C-order index.
- `'f_index'`: Track the Fortran-order index.
- `'multi_index'`: Track a multi-dimensional index.
- `'external_loop'`: Return multiple elements at once in a 1D array.
- `order`: Specifies the traversal order ('C', 'F', or 'A' for adaptive).
- `op_dtypes`: Specifies the data type for the iterator.

nditer is a versatile tool, but for simpler cases, basic Python iteration or array operations may be sufficient. Use nditer when you need fine-grained control over iteration or performance optimization.

```python
arr = np.array([1, 2, 3], dtype=np.int32)
for x in np.nditer(arr, flags=['refs_ok'], op_dtypes=[np.float64]):
    print(x)
```

```python
import numpy as np

# Create a NumPy array
arr = np.array([1, 2, 3, 4])

# Use nditer with write access
for i in np.nditer(arr, op_flags=['readwrite']):
    i[...] = i * 2  # Modify the value in place

print(arr)
```
