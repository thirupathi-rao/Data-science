```python
#1 What is the average weight of Pokémon in each generation?
```

```python
df.groupby('generation')['weight'].mean()
```

```python
ax=sns.barplot(data=df,x='generation',y='weight',errorbar=None)
for i in ax.containers:
    ax.bar_label(i,)
```
